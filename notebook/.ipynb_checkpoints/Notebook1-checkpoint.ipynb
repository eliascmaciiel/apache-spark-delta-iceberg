```python
import sys
import os

sys.path.append(os.path.abspath("../src"))

from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("DeltaExample")
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")
    .getOrCreate()
)
```

```python
data = [
    (1, "João", "SP"),
    (2, "Maria", "RJ")
]

df = spark.createDataFrame(data, ["id", "nome", "cidade"])
df.show()
```

```python
df.write.format("delta").mode("overwrite").save("data/delta/clientes")
```

```python
df = spark.read.format("delta").load("data/delta/clientes")
df.show()
```

```python
spark.sql("""
INSERT INTO delta.`data/delta/clientes`
VALUES (3, 'Carlos', 'SC')
""")

spark.read.format("delta").load("data/delta/clientes").show()
```

```python
spark.sql("""
UPDATE delta.`data/delta/clientes`
SET cidade = 'PR'
WHERE id = 1
""")

spark.read.format("delta").load("data/delta/clientes").show()
```

```python
spark.sql("""
DELETE FROM delta.`data/delta/clientes`
WHERE id = 2
""")

spark.read.format("delta").load("data/delta/clientes").show()
```
